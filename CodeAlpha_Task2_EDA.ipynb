```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv("CodeAlpha_Task1_Final_Dataset.csv")

print("Dataset loaded successfully!")
```

```python
df.describe()
```

```python
print("Average Price: £", round(df["Price"].mean(), 2))
print("Average Rating:", round(df["Rating"].mean(), 2))
```

```python
cheapest_book = df.loc[df["Price"].idxmin()]
expensive_book = df.loc[df["Price"].idxmax()]

print("Cheapest Book:")
print(cheapest_book)

print("\nMost Expensive Book:")
print(expensive_book)
```

```python
rating_counts = df["Rating"].value_counts().sort_index()

print(rating_counts)
```

```python
plt.figure(figsize=(8, 5))

plt.hist(df["Price"], bins=10)

plt.xlabel("Book Price (£)")
plt.ylabel("Number of Books")
plt.title("Distribution of Book Prices")

plt.show()
```

```python
rating_counts = df["Rating"].value_counts().sort_index()

plt.figure(figsize=(8, 5))

plt.bar(rating_counts.index, rating_counts.values)

plt.xlabel("Rating")
plt.ylabel("Number of Books")
plt.title("Number of Books by Rating")

plt.xticks([1, 2, 3, 4, 5])

plt.show()
```

```python
plt.figure(figsize=(8, 5))

plt.scatter(df["Rating"], df["Price"])

plt.xlabel("Rating")
plt.ylabel("Book Price (£)")
plt.title("Relationship Between Book Rating and Price")

plt.show()
```

```python
eda_summary = pd.DataFrame({
    "Metric": [
        "Total Books",
        "Average Price (£)",
        "Minimum Price (£)",
        "Maximum Price (£)",
        "Average Rating"
    ],
    "Value": [
        len(df),
        round(df["Price"].mean(), 2),
        round(df["Price"].min(), 2),
        round(df["Price"].max(), 2),
        round(df["Rating"].mean(), 2)
    ]
})

eda_summary
```
